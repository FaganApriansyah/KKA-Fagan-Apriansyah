```python
class Hero:
 # Constructor: Dijalankan saat Hero baru dibuat
 def __init__(self, name, hp, attack_power):
  self.name = name # Nama Hero
  self.hp = hp # Nyawa (Health Point)
  self.attack_power = attack_power # Kekuatan Serangan

 # Method untuk menampilkan info hero
 def info(self):
   print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

 # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
 def serang(self, lawan):
  print(f"{self.name} menyerang {lawan.name}!")
  lawan.diserang(self.attack_power)

 # Method diserang: Menerima damage
 def diserang(self, damage):
  self.hp -= damage
  print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# Class Mage adalah anak dari class Hero
class Mage(Hero):
 def __init__(self, name, hp, attack_power, mana):

 # Memanggil constructor milik Parent (Hero)
  super().__init__(name, hp, attack_power)
  self.mana = mana # Corrected indentation
 def info(self):
  print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

 # Mage punya skill khusus
 def skill_fireball(self, lawan):
  if self.mana >= 20:
   print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
   self.mana -= 20
   lawan.diserang(self.attack_power * 2) # Damage 2x lipat
  else:
   print(f"{self.name} gagal skill! Mana tidak cukup.")

# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)
eudora.info()
eudora.serang(balmond) # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond) # Skill khusus Mage

# -- Main Program --

# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero1.hp = 500
print(hero1.hp)
hero2 = Hero("Zilong", 120, 20)

# Memanggil Method
hero1.info()
hero2.info()

 # Tambah kode Output di akhir program
print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2) # Layla menyerang Zilong
hero2.serang(hero1) # Zilong membalas
```

```python
class Hero:
 def __init__(self, nama, hp_awal):
  self.nama = nama
  # Enkapsulasi: HP bersifat Private
  self.__hp = hp_awal

 # GETTER
 def get_hp(self):
  return self.__hp

 # SETTER dengan validasi
 def set_hp(self, nilai_baru):
  if nilai_baru < 0:
   self.__hp = 0
  elif nilai_baru > 1000:
   print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
   self.__hp = 1000
  else:
   self.__hp = nilai_baru

 def diserang(self, damage):
  # Memakai setter/getter agar logika validasi tetap jalan
  sisa_hp = self.get_hp() - damage
  self.set_hp(sisa_hp)
  print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
hero1 = Hero("Layla", 100)
hero1.set_hp(-50) # Akan diset ke 0 oleh setter
print(f"HP Setelah set negatif: {hero1.get_hp()}")

hero1.diserang(20)
print(f"HP Setelah diserang: {hero1.get_hp()}")
```

Selain itu, cara ini membuat kode menjadi sangat ringkih; apabila nama variabel internal diubah di dalam class, kode luar yang mengaksesnya secara paksa akan langsung mengalami crash (AttributeError).

```python
from abc import ABC, abstractmethod
# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.
class GameUnit(ABC):

 @abstractmethod
 def serang(self, target):
  pass

 @abstractmethod
 def info(self):
  pass

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
 def __init__(self, nama):
  self.nama = nama

 # Kita WAJIB membuat method serang, kalau tidak akan Error
 def serang(self, target):
  print(f"Hero {self.nama} menebas {target}!")

 def info(self):
  print(f"Saya adalah Hero: {self.nama}")

class Monster(GameUnit):
 def __init__(self, jenis):
  self.jenis = jenis

 # Implementasi serang versi Monster (Sekarang sudah di-indent dengan benar)
 def serang(self, target):
  print(f"Monster {self.jenis} menggigit {target}!")

 def info(self):
  print(f"Saya adalah Monster: {self.jenis}")

# -- Uji Coba --
h = Hero("Alucard")
m = Monster("Serigala")
h.info()
m.info()
```

Menghapus method serang() memicu TypeError karena class Hero dianggap belum melengkapi kontrak wajib dari abstract class. Konsekuensinya, program langsung terhenti (crash) saat objek dibuat guna mencegah eror akibat pemanggilan fungsi yang tidak ada. Class GameUnit sendiri dilarang diinstansiasi langsung karena hanyalah konsep rancangan tanpa wujud nyata dalam game. Keberadaannya berfungsi sebagai cetak biru atau kontrak standar yang menjamin seluruh class anak memiliki struktur method yang seragam.

```python
# Parent Class
class Hero:
 def __init__(self, nama):
  self.nama = nama

 def serang(self):
  print("Hero menyerang dengan tangan kosong.")
# Child Class 1
class Mage(Hero):
 def serang(self):
  print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")
# Child Class 2
class Archer(Hero):
  def serang(self):
   print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")
# Child Class 3
class Fighter(Hero):
 def serang(self):
  print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")
# -- Penerapan Polymorphism --
# Kita punya daftar hero campuran
pasukan = [
 Mage("Eudora"),
 Archer("Miya"),
 Fighter("Zilong"),
 Mage("Gord")
]
print("--- PERANG DIMULAI ---")
# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
 pahlawan.serang()

```

Kesimpulan: Nama method harus sama agar satu perintah dapat digunakan untuk berbagai objek dengan hasil yang berbeda.

```python
from abc import ABC, abstractmethod


# --- 1. abstraksi and enkapsulasi (parent class) ---
class KamarHotel(ABC):

    def __init__(self, nama_kamar, harga_dasar):
        self.nama_kamar = nama_kamar
        self.__harga_dasar = harga_dasar  # Private attribute
        self.__stok = 0  # Private attribute

    # Getter stok
    def get_stok(self):
        return self.__stok

    # Setter & Validasi stok
    def tambah_stok(self, jumlah):
        # Mengambil label singkat (misal: "Kamar Deluxe") untuk tampilan log
        label_singkat = " ".join(self.nama_kamar.split()[:2])
        if jumlah < 0:
            print(
                f"Gagal update stok {label_singkat}! Stok tidak boleh negatif ({jumlah})."
            )
        else:
            self.__stok += jumlah
            print(
                f"Berhasil menambahkan stok {label_singkat}: {jumlah} unit."
            )

    # Getter harga dasar
    def get_harga_dasar(self):
        return self.__harga_dasar

    # Abstract Methods (Wajib di-override oleh class anak)
    @abstractmethod
    def tampilkan_detail(self, jumlah_malam):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


# --- 2. INHERITANCE & POLYMORPHISM (CHILD CLASSES) ---
class KamarDeluxe(KamarHotel):

    def __init__(self, nama_kamar, harga_dasar, fasilitas):
        super().__init__(nama_kamar, harga_dasar)
        self.fasilitas = fasilitas
        self.persen_pajak = 0.10  # Pajak 10%

    def hitung_pajak(self):
        return self.get_harga_dasar() * self.persen_pajak

    def hitung_harga_total(self, jumlah_malam):
        harga_per_malam = self.get_harga_dasar() + self.hitung_pajak()
        return harga_per_malam * jumlah_malam

    def tampilkan_detail(self, jumlah_malam):
        pajak = self.hitung_pajak()
        subtotal = self.hitung_harga_total(jumlah_malam)
        print(
            f"[DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}\n"
            f" Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f}".replace(
                ",", "."
            )
            + f" | Pajak(10%): Rp {pajak:,.0f}\n".replace(",", ".")
            + f" Menginap: {jumlah_malam} malam | Subtotal: Rp {subtotal:,.0f}".replace(
                ",", "."
            )
        )


class KamarStandard(KamarHotel):

    def __init__(self, nama_kamar, harga_dasar, kapasitas):
        super().__init__(nama_kamar, harga_dasar)
        self.kapasitas = kapasitas
        self.persen_pajak = 0.05  # Pajak 5%

    def hitung_pajak(self):
        return self.get_harga_dasar() * self.persen_pajak

    def hitung_harga_total(self, jumlah_malam):
        harga_per_malam = self.get_harga_dasar() + self.hitung_pajak()
        return harga_per_malam * jumlah_malam

    def tampilkan_detail(self, jumlah_malam):
        pajak = self.hitung_pajak()
        subtotal = self.hitung_harga_total(jumlah_malam)
        print(
            f"[STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}\n"
            f" Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f}".replace(
                ",", "."
            )
            + f" | Pajak(5%): Rp {pajak:,.0f}\n".replace(",", ".")
            + f" Menginap: {jumlah_malam} malam | Subtotal: Rp {subtotal:,.0f}".replace(
                ",", "."
            )
        )


# --- 3. FITUR PEMESANAN (POLYMORPHISM OUTSIDE CLASS) ---
def proses_transaksi(daftar_pesanan):
    print("--- STRUK PEMESANAN ---")
    total_tagihan = 0

    for idx, (kamar, malam) in enumerate(daftar_pesanan, start=1):
        print(f"{idx}. ", end="")
        kamar.tampilkan_detail(malam)
        total_tagihan += kamar.hitung_harga_total(malam)

    print("-" * 40)
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}".replace(",", "."))
    print("-" * 40)


# --- 4. RUNNER PROGRAM (USER STORY) ---
if __name__ == "__main__":
    # a) Admin membuat data kamar
    kamar_deluxe = KamarDeluxe(
        "Kamar Deluxe Sea View", 1500000, "Private Pool"
    )
    kamar_standard = KamarStandard(
        "Kamar Standard Superior", 500000, "2 Orang"
    )

    # b) Pengujian validasi stok
    print("--- SETUP DATA KAMAR ---")
    kamar_deluxe.tambah_stok(10)
    kamar_standard.tambah_stok(-5)  # Ditolak karena negatif
    kamar_standard.tambah_stok(20)

    # c & d) Tamu memesan kamar & Cetak Struk
    pesanan_tamu = [
        (kamar_deluxe, 2),  # 2 malam Deluxe
        (kamar_standard, 1),  # 1 malam Standard
    ]

    proses_transaksi(pesanan_tamu)
```
